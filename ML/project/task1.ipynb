#### Step 4: Convert Age from Days to Years

```python
df['age_years'] = (df['age'] / 365).astype(int)
print(df['age_years'].describe())
```

#### Step 5: Remove Unrealistic Height and Weight Outliers

```python
df = df[(df['height'] >= 120) & (df['height'] <= 220)]

df = df[(df['weight'] >= 30) & (df['weight'] <= 250)]
```

#### Step 6: Fix Blood Pressure (ap_hi, ap_lo) Outliers

```python
df = df[(df['ap_hi'] >= 90) & (df['ap_hi'] <= 200)]
df = df[(df['ap_lo'] >= 60) & (df['ap_lo'] <= 120)]
```

#### Step 7: Calculate BMI 

```python
df['BMI'] = df['weight'] / ((df['height']/100) ** 2)
df
```
