# Perform conditional selection to find - Population ages 15-64 (% of total population)

```python
data = df[df['Indicator Name'] == "Population ages 15-64 (% of total population)"]
data
```

# Divide the data into input and output

```python
import numpy as np
```

```python
X = np.arange(1960, 2020).reshape(-1,1)
X
```

```python
y = data.values[0][4:-1].reshape(-1,1)
y
```

# Plot scatter plot of Population ages 15-64 (% of total population)

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(X,y)
plt.show()
```

# Feature Scaling (Mandatory for SVR)**
SVR is highly sensitive to the range of data points. If we don't scale (normalize) the data, the model will fail to find the correct hyperplane.

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc_X = StandardScaler()
sc_y = StandardScaler()

x_scale = sc_X.fit_transform(X)
y_scale = sc_y.fit_transform(y)

x_scale
```

```python
y_scale
```

# Fitting SVR on Linear Kernel on dataset

```python
from sklearn.svm import SVR
```

```python
svr_lin = SVR(kernel="linear")
```

```python
svr_ply = SVR(kernel="rbf", degree=5)
```

```python
svr_lin.fit(x_scale, y_scale.ravel())
```

```python
svr_ply.fit(x_scale, y_scale.ravel())
```

# Predict Values using Model

```python
# use inverse_transform
```

```python
y_poly = svr_ply.predict(x_scale).reshape(-1,1)
y_poly
```

```python
y_pred = svr_lin.predict(x_scale).reshape(-1,1)
y_pred
```

```python
orignal_y_poly = sc_y.inverse_transform(y_poly)
```

```python
orginal_y = sc_y.inverse_transform(y_pred)
orginal_y
```

```python
plt.scatter(X,y)
plt.scatter(X,orignal_y_poly)
```

```python
plt.scatter(X,y)
plt.scatter(X,orginal_y)
```
